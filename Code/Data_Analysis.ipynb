```python
Default_df = df[df['12-Month Default Flag'] == 1]
Normal_df = df[df['12-Month Default Flag'] == 0]
```

```python
def Correlation_plot(df):

    corr = df.select_dtypes(
        include='number',
    ).corr(method='spearman')

    plt.figure(figsize=(20, 10))
    sns.heatmap(
        corr,
        annot=True,
        cmap='coolwarm',
        center=0,
        fmt='.2f'
    )

    plt.title("Spearman Correlation Matrix")
    plt.tight_layout()
    plt.show()
```

```python
Correlation_plot(Default_df)
```

```python
Correlation_plot(Normal_df)
```
